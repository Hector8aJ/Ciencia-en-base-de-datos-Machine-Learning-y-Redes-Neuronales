```python
import tensorflow as tf
import numpy as np
import pandas as pd
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
import numpy as np
```

```python
# Definimos un conjunto de frases de ejemplo y sus etiquetas correspondientes
texts = [
"Me encanta el deep learning",
"Odio las matemáticas",
"El cine es increíble",
"La política es aburrida",
"La música es maravillosa",
"No soporto el tráfico",
"Los videojuegos son divertidos",
"El calor extremo es insoportable"
]
labels = [1, 0, 1, 0, 1, 0, 1, 0] # 1: Positivo, 0: Negativo
```

```python
vocab_size = 1000 # Número máximo de palabras en el vocabulario
tokenizer = Tokenizer(num_words=vocab_size, oov_token="<OOV>") # OOV para palabras desconocidas
tokenizer.fit_on_texts(texts)
word_index = tokenizer.word_index
sequences = tokenizer.texts_to_sequences(texts) # Convertimos las frases en secuencias numéricas
```

```python
# Padding: Aseguramos que todas las secuencias tengan la misma longitud
max_length = 10 # Longitud máxima de cada secuencia
data = pad_sequences(sequences, maxlen=max_length, padding='post', truncating='post')
```

```python
# Construcción del modelo de red neuronal con Embeddings y LSTM
model = Sequential([
Embedding(input_dim=vocab_size, output_dim=64, input_length=max_length), # Capa de embedding
LSTM(64, return_sequences=True), # Capa LSTM con salida en cada paso de tiempo
LSTM(32), # Segunda capa LSTM para extraer más información
Dense(16, activation='relu'), # Capa densa intermedia
Dense(1, activation='sigmoid') # Capa de salida con activación sigmoide para clasificación binaria
])
```

```python
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

```python
history = model.fit(data, np.array(labels), epochs=30, verbose=1)
```

```python
loss, accuracy = model.evaluate(data, np.array(labels))
print(f"Pérdida en entrenamiento: {loss:.4f}")
print(f"Precisión en entrenamiento: {accuracy:.4f}")
```

```python
# Prueba con nuevas frases
test_texts = ["Me gusta chuparle la cola a Andrea", "Odio a los nazis"]
test_sequences = tokenizer.texts_to_sequences(test_texts)
test_data = pad_sequences(test_sequences, maxlen=max_length, padding='post', truncating='post')
```

```python
# Predicción
predictions = model.predict(test_data)
print("\nPredicciones:")

for text, pred in zip(test_texts, predictions):  # Asegúrate de que haya una indentación aquí
    print(f"Texto: '{text}' -> Sentimiento: {'Positivo' if pred > 0.5 else 'Negativo'} ({pred[0]:.4f})")
```
