# Regresión logística múltiple Heart


This data set dates from 1988 and consists of four databases: Cleveland, Hungary, Switzerland, and Long Beach V. It contains 76 attributes, including the predicted attribute, but all published experiments refer to using a subset of 14 of them. The "target" field refers to the presence of heart disease in the patient. It is integer valued 0 = no disease and 1 = disease

[https://www.kaggle.com/datasets/johnsmith88/heart-disease-dataset](https://www.kaggle.com/datasets/johnsmith88/heart-disease-dataset)

```python
# División de los datos en train y test
# ==============================================================================
X = df.drop(columns= 'target')
y = df['target']
X_train, X_test, y_train, y_test = train_test_split(
                                        X,
                                        y.values.reshape(-1,1),
                                        train_size   = 0.8,
                                        random_state = 123,
                                        shuffle      = True
                                    )
```

```python
X.head()
```

```python
# Creación del modelo utilizando matrices como en scikitlearn
# ==============================================================================
# A la matriz de predictores se le tiene que añadir una columna de 1s para el intercept del modelo
X_train = sm.add_constant(X_train, prepend=True)
modelo = sm.Logit(endog = y_train, exog = X_train)
modelo_fit = modelo.fit()
print(modelo_fit.summary())
```

## Predicciones en el conjunto de entrenamiento

```python
# Predicciones con intervalo de confianza
# ==============================================================================
predicciones = modelo_fit.predict(exog= X_train)

# Clasificación predicha
# ==============================================================================
clasificacion = np.where(predicciones<0.5, 0, 1)
clasificacion
```

```python
# Matriz de confusión de las predicciones de train
# ==============================================================================
cm = pd.crosstab(
    y_train.ravel(),
    clasificacion,
    rownames=['Real'],
    colnames=['Prediccion']
)
cm
```

```python
accuracy = accuracy_score(
    y_true= y_train,
    y_pred= clasificacion,
    normalize= True
)
print("")
print(f"El accuracy de train es : {100*accuracy}%")
```

## Predicciones sobre el cojunto de prueba

Se calcula el porcentaje de aciertos que tiene el modelo al predecir las observaciones de test (accuracy).

```python
X_test = sm.add_constant(X_test, prepend=True) #Agregamos la constante a x_test
pred_test = modelo_fit.predict(exog = X_test) #Realiza las predicciones en el conjunto de prueba

pred_test = np.where(pred_test<0.5, 0,1)
```

```python
# Accuracy de test del modelo
# ==============================================================================

accuracy = accuracy_score(
            y_true    = y_test,
            y_pred    = pred_test,
            normalize = True
           )
print("")
print(f"El accuracy de test es: {100*accuracy}%")
```

### Matriz de confusión

```python
# Matriz de confusión de las predicciones de test
# ==============================================================================
cm1 = pd.crosstab(
    y_test.ravel(),
    pred_test,
    rownames=['Real'],
    colnames=['Prediccion']
)
cm1
```

```python
 ## Metreicas de evaluación
accuracy = accuracy_score(
            y_true    = y_test,
            y_pred    = pred_test,
            normalize = True
           )
print("")
print(f"El accuracy de test es: {100*accuracy}%")

accuracy = accuracy_score(y_test, pred_test)
precision = precision_score(y_test, pred_test)
recall = recall_score(y_test, pred_test)
f1 = f1_score(y_test, pred_test)
auc_roc = roc_auc_score(y_test, pred_test)

# Imprimir métricas
print("Exactitud (Accuracy):", accuracy)
print("Precisión (Precision):", precision)
print("Recall (Sensitivity):", recall)
print("F1-score:", f1)
print("AUC-ROC:", auc_roc)
```
