```python
import tensorflow as tf
import numpy as np

# Generar datos de ejemplo (regresión lineal)
np.random.seed(42)
X = np.random.rand(100, 1)  # 100 puntos en 1D
y = 3 * X + 4 + 0.1 * np.random.randn(100, 1)  # y = 3x + 4 + ruido

# Crear un modelo lineal simple
def create_model():
    model = tf.keras.Sequential([
        tf.keras.layers.Dense(1, input_shape=(1,))  # Una sola neurona para regresión lineal
    ])
    return model

# Compilar el modelo con Adam
model = create_model()
model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=0.01, beta_1=0.9, beta_2=0.99),  #BETA 1 factor de decaimiento #Factor de decaimiento 2º momento
              loss='mse')  # Usamos el error cuadrático medio (MSE)

# Entrenar el modelo
history = model.fit(X, y, epochs=100, verbose=0) ##No muestra el progreso del entrenamiento si está el verbose=0

# Graficar la pérdida durante el entrenamiento
import matplotlib.pyplot as plt
plt.plot(history.history['loss'])
plt.xlabel('Época')
plt.ylabel('Pérdida (MSE)')
plt.title('Pérdida durante el entrenamiento con Adam')
plt.show()
```
